```python
from scipy.special import gammaincc
from scipy.special import gamma
d=2
m=1

def Gam(a,x):
    return gammaincc(a,x)*gamma(a)
def Gam0(a):
    return gamma(a)
ketgam=(Gam(5/2+d/m,5)/Gam0(5/2+d/m))/(Gam(3/2+d/m,5)/Gam0(3/2+d/m)-1)-1
ketgam2=(Gam(5/2+d/m,5)/Gam0(5/2+d/m))/(Gam(3/2+d/m,5)/Gam0(3/2+d/m))-1

```
